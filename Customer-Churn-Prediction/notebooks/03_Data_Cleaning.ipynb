```python
import pandas as pd

# Step 1: Load the original dataset
df = pd.read_csv("../dataset/Telco-Customer-Churn.csv")

print("Original shape:", df.shape)

# Step 2: Remove extra spaces from TotalCharges
df["TotalCharges"] = df["TotalCharges"].astype(str).str.strip()

# Step 3: Convert TotalCharges into numeric format
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

print("Missing TotalCharges after conversion:",
      df["TotalCharges"].isnull().sum())

# Step 4: Fill missing TotalCharges with 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)

print("Shape after filling missing values:", df.shape)

# Step 5: Drop customerID
df = df.drop(columns=["customerID"])

print("Shape after dropping customerID:", df.shape)

# Step 6: Verify the dataset
print("\nFinal shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\nChurn distribution:")
print(df["Churn"].value_counts())

# Step 7: Save the corrected dataset
df.to_csv(
    "../dataset/Telco-Customer-Churn-Cleaned-v2.csv",
    index=False
)

print("\nCorrected dataset saved successfully!")
```
